```python
def get_validation_throughout_evolution(dict, gen_max, populations=False, multitask=False):
    """_summary_

    Args:
        dict (_type_): _description_
        gen_max (_type_): _description_
        populations (bool, optional): _description_. Defaults to False.
        multitask (bool, optional): _description_. Defaults to False.

    Returns:
        _type_: _description_
    """
    # Validation scores are saved in a matrix with dimensions:
    # generations x hyperparameter candidates x re-initializations x lag search grid
    if not multitask:
        all_scores = dict['validation performance'][:gen_max, :, :, :]
    else:
        all_scores = dict['validation performance'][:gen_max, :, :, :, :]
    # from the lag search grid we select the best score (lowest NRMSE), since this was the best 
    # performing readout model
    best_lag_scores = np.min(all_scores, axis=-1)
    
    if multitask:
        # from the tasks, we select the average
        # best_lag_scores = best_lag_scores[:, :, :, 0]**2
        best_lag_scores = np.mean(best_lag_scores, axis=-1)
    
    # from the re-initializations from the same hyperparameter set/candidate, we take the average
    best_candidate_scores = np.mean(best_lag_scores, axis=-1)
    
    if populations:
        return best_candidate_scores
    
    # from the population of hyperparameter candidates, we select the best candidate
    best_gen_scores = np.min(best_candidate_scores, axis=-1)
    return best_gen_scores

def get_best_candidate(dict, gen_max, multitask=False):
    """_summary_

    Args:
        dict (_type_): _description_
        gen_max (_type_): _description_
        multitask (bool, optional): _description_. Defaults to False.

    Returns:
        _type_: _description_
    """
    best_gen_scores = get_validation_throughout_evolution(dict, gen_max, multitask=multitask)   
    best_candidates = get_validation_throughout_evolution(dict, gen_max, populations=True, multitask=multitask)
    best_gen_ind = np.argmin(best_gen_scores)
    best_candidate_ind = np.argmin(best_candidates[best_gen_ind])
    return best_gen_ind, best_candidate_ind
    
def sample_best_net(dict, n, gen_max, multitask=False):
    """_summary_

    Args:
        dict (_type_): _description_
        n (_type_): _description_
        gen_max (_type_): _description_
        multitask (bool, optional): _description_. Defaults to False.

    Returns:
        _type_: _description_
    """
    gen, ind = get_best_candidate(dict, gen_max, multitask=multitask)
    print(gen, ind)
    start_net = dict['example net']
    best_parameter_set = dict['parameters'][gen, ind, :]
    best_nets = []
    for i in range(n):
        best_net = start_net.get_new_network_from_serialized(best_parameter_set)
        best_nets.append(best_net)
    return best_nets
```

```python
def summarize_stabilization(
    act: np.ndarray,
    ts_plot: np.ndarray,
    mask_plot: np.ndarray,
    switch_idx: int,
    title: str,
    tau_axis: np.ndarray = None,
    rho: float = None,
    x_star: np.ndarray = None,   # NEW: the actual fixed point, if you have it
    W=None,
    b=None,
    t_id: int = None,
):
    fig, axes = plt.subplots(
        2, 3,
        figsize=(14, 5),
        gridspec_kw={"width_ratios": [1.25, 1.25, 1.0], "wspace": 0.35, "hspace": 0.5},
    )

    act = np.asarray(act, dtype=float)
    N, T = act.shape

    # --- Panel 1: activity (unchanged) ---
    for k in (0, 1, 2):
        axes[0, k].plot(ts_plot, act[k], lw=0.9, alpha=0.7)
    axes[0, k].axvline(ts_plot[switch_idx], color="red", lw=1.2, ls="--")
    axes[0, k].set_title(title, loc="right", fontsize=10)
    axes[0, k].set_ylabel("unit activation")

    # --- Panel 4: one-step change + fixed-point residual ---
    delta_x = np.linalg.norm(np.diff(act, axis=1), axis=0)

    axes[1, 0].semilogy(ts_plot[1:], delta_x, lw=0.8, color="teal",
                        label=r"$\|x_{n+1}-x_n\|$")
    axes[1, 0].axvline(ts_plot[switch_idx], color="red", lw=1.2, ls="--")
    axes[1, 0].set_ylabel(r"step size $\|x_{n+1}-x_n\|$")
    axes[1, 0].legend(loc="upper right", fontsize=8)
    axes[1, 0].set_ylim(1e-16, None)

    # If you have the actual fixed point, also plot the residual to it.
    if x_star is not None:
        x_star = np.asarray(x_star, dtype=float).reshape(-1)
        if x_star.size != N:
            raise ValueError(f"x_star has size {x_star.size}; expected {N}.")
        dist_x_star = np.linalg.norm(act - x_star[:, None], axis=0)
        axes[1, 0].plot(ts_plot, dist_x_star, lw=0.8, color="purple",
                        alpha=0.6, label=r"$\|x_n - x^\star\|$")
        axes[1, 0].legend(loc="upper right", fontsize=8)

    # --- Panel 5: distance to final state + cycle check ---
    x_final = act[:, -1]
    dist_x_final = np.linalg.norm(act - x_final[:, None], axis=0)

    axes[1, 1].semilogy(ts_plot, dist_x_final, lw=0.8, color="steelblue",
                        label=r"$\|x_n - x_{\mathrm{final}}\|$")
    if rho is not None:
        # Anchor the rho line to the distance AT the switch, not at n=0.
        n_after = np.arange(len(ts_plot) - switch_idx)
        axes[1, 1].plot(
            ts_plot[switch_idx:], dist_x_final[switch_idx] * rho**n_after,
            lw=1.2, ls="--", color="firebrick", label=r"$\rho^{n}$",
        )
    axes[1, 1].axvline(ts_plot[switch_idx], color="red", lw=1.2, ls="--")
    axes[1, 1].set_ylabel(r"$\|x_n - x_{\mathrm{final}}\|$")
    axes[1, 1].legend(loc="upper right", fontsize=8)
    axes[1, 1].set_ylim(1e-16, None)

    # Cycle check: is the network on a periodic orbit after the switch?
    free_len = T - switch_idx
    period_checks = {}
    for p in range(1, min(5, free_len // 2)):
        a = act[:, switch_idx:-p]
        b_ = act[:, switch_idx + p:]
        period_checks[p] = float(np.max(np.linalg.norm(a - b_, axis=0)))
    period_str = ", ".join(f"p={p}:{d:.1e}" for p, d in period_checks.items())

    axes[1, 1].text(
        0.02, 0.05,
        f"period check: {period_str}",
        transform=axes[1, 1].transAxes, fontsize=7,
        bbox=dict(facecolor="white", alpha=0.7),
    )

    # --- Panel 6: PCA (fit on driven segment) ---
    driven = act[:, mask_plot == 1]
    pca = PCA(n_components=2).fit(driven.T)
    pc = pca.transform(act.T)

    axes[1, 2].plot(pc[mask_plot == 1, 0], pc[mask_plot == 1, 1],
                   lw=0.7, alpha=0.7)
    axes[1, 2].plot(pc[mask_plot == 0, 0], pc[mask_plot == 0, 1],
                   color="darkgreen", lw=0.7, alpha=0.9)
    axes[1, 2].scatter(0, 0, marker="x", c="red", s=60, label="end")
    axes[1, 2].text(
        0.98, 0.05, f"$\\|x_{{end}}\\| = {np.linalg.norm(act[:, -1]):.2f}$",
        transform=axes[1, 2].transAxes, ha="right", fontsize=9,
    )
    axes[1, 2].set_title("PCA (fit on driven segment)", loc="left", fontsize=10)
    axes[1, 2].set_xlabel("PC1")
    axes[1, 2].set_ylabel("PC2")
    axes[1, 2].legend(fontsize=8)

    return fig
```

```python
def generate_mask(ts_plot, switch_idx):
    mask = np.zeros_like(ts_plot, dtype=int)
    mask[:switch_idx] = 1  # driven segment
    mask[switch_idx:] = 0  # free segment
    return mask
```
